## Download dos dados

Os dados utilizados neste repositório foram disponibilizados em https://openneuro.org/datasets/ds003626. Este notebook faz download desses dados no diretório `data/ds003626/derivatives`.

```python
# Imports
from pathlib import Path
import sys

# Add the Inner Speech Dataset repository to the system path
sys.path.append(str(Path.cwd().parent / "src" / "Python_Processing"))

# Import custom functions from the Inner Speech Dataset repository
from lib.download_helper import get_derivatives
```

```python
# List with every subject (S1-S10)
N_S = list(range(1, 11))

# sub-09 ses-01 and sub-10 ses-03 are truncated in the OpenNeuro datasets
EXCLUDE = {9, 10}
SUBJECTS = [i for i in N_S if i not in EXCLUDE]

root_dir = Path.cwd().parent / "data" / "ds003626"

for i in SUBJECTS:
    get_derivatives(
        dataset_path=root_dir,
        subjects=f"{i:02d}",
        sessions=None,
        file_types=["eeg","exg","events"],
        verbose=True,
        progress=True,
    )
```
